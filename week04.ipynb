## Path graph

We looked at the maximum-weight independent set **(MWIS)** for a path graph. A path graph $G_n=(V,E,\mathbf w)$ has $n$ vertices $\{v_1, v_2, \ldots, v_n\}$. Each vertex has a weight $w_i$ assigned to it. There are $n-1$ edges in a path graph:

$$
E = \{\, (v_1,v_2),\, (v_2, v_3),\, \ldots,\, (v_{n-2}, v_{n-1}),\, (v_{n-1}, v_n)\, \}
$$

## MWIS for a path graph

The MWIS of $G_n$ is a set $S\subseteq V$ such that

$$ \forall u,v\in S,\, (u,v) \not\in V$$
and
$$\arg\max_{S} \sum_{u\in S} w_u$$

In other words, we are looking for the heaviest subset of vertices that are not adjacent to each other. A very simple example is a graph with equal weight vertices, i.e., $w_1=w_2=\ldots=w_n$. In this case, $S=\{v_1, v_3, \ldots\}$. In general, however, the set $S$ is derived based on an obvious  observation:
**either** $v_n\not\in S$ **or** $v_n\in S$.

We've shown in class that if $v_n\not\in S$ then $S$ is also the MWIS for $G_{n-1}$. And when $v_n\in S$, then $S$ is also the MWIS for $G_{n-2}\cup \{v_n\}$.

Let $W_n$ be the weight for the MWIS for $G_n$:

$$
W_n = \sum_{u\in S(G_n)} w_u
$$

We showed that 

$$ W_n = \max\{W_{n-1}, W_{n-2}+w_n\} $$
with initial conditions $W_0 = 0$ and $W_1 = w_1$. The corresponding MWIS are $S(G_0)=\varnothing$ and $S(G_1)=v_1$. Reconstruction of $S(G_i)$ for $i>1$ require a bit more effort.
